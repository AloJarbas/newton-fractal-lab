# Asymmetric cubic geometry contrast

This notebook is the slow companion to the new sidecar in `art/asymmetric-cubic-geometry-contrast.*`.

The narrow question is worth slowing down:

**once the unity-family symmetry is already broken, do different asymmetric cubics still tell the same Newton story?**

The answer in this repo is no.


![Asymmetric cubic geometry contrast](../art/asymmetric-cubic-geometry-contrast.png)

## Scope and boundary

This notebook does **not** try to classify all cubic Newton fractals.

It compares just two bounded asymmetric examples:

1. the repo's earlier asymmetric cubic, which quickly becomes a winner-take-most square,
2. a new split-critical asymmetric cubic, whose critical points stay much closer to the center and keep the basin competition alive.

That is enough to test whether “broken symmetry” by itself already predicts the geometry. It does not.


## The two cubics

### Existing asymmetric cubic

\[
p_a(z) = (z-1)(z-(-0.9+1.05i))(z-(-0.15-0.25i)).
\]

### Split-critical asymmetric cubic

\[
p_s(z) = (z-1)(z-(-0.35+0.92i))(z-(-0.30-0.88i)).
\]

The important difference is not only the roots.
It is the critical geometry induced by those roots.

For a cubic

\[
p(z) = z^3 + az^2 + bz + c,
\]

the Newton tension is organized by the zeros of

\[
p'(z) = 3z^2 + 2az + b.
\]

Those are the critical points used in the distance-band scan.


## Main measurements from the shipped sidecar

Using the same sampled square and Newton budget as the figure:

| quantity | existing asymmetric cubic | split-critical asymmetric cubic |
|---|---:|---:|
| largest basin share on the square | `57.9%` | `38.6%` |
| smallest basin share on the square | `16.4%` | `29.5%` |
| nearest-critical late fraction | `12.2%` | `36.0%` |
| outer-band dominant share | `100.0%` | `59.9%` |

That table is the whole point of the pass.


## What changed in the story

### 1. The first asymmetric cubic was not “the” asymmetric outcome

The earlier asymmetric sidecar taught one real lesson:

- once symmetry breaks, one root can start owning much more of the sampled square.

That lesson survives.
It is just not universal.

### 2. The split-critical cubic keeps the competition alive

The new cubic breaks symmetry too.
But it does **not** collapse into the same basin-share story.

Its largest basin only reaches about `38.6%`, and the smallest still keeps about `29.5%`.
So the sampled square stays much closer to a genuine three-way competition.

### 3. The near-critical lane stays hotter

The more surprising difference is not only the basin shares.
It is the near-critical late tail.

The existing asymmetric cubic cools that band down to about `12.2%`.
The split-critical cubic keeps it at about `36.0%`.

So the new example is not just “another asymmetric cubic.”
It preserves a much more active critical zone.


## Why distance to the nearest critical point is still the right coordinate here

The unity cubic let the origin stand in for the critical set because the critical point was repeated there.

These asymmetric cubics remove that shortcut.
The nearest-critical distance remains the clean local coordinate because it asks the right question for both examples:

- how close is the start to the region where Newton's denominator becomes fragile,
- and how fast does that fragility cool as you move away?

That is why the figure uses distance-band scans instead of only another basin poster.


## Caveats

- This is still a bounded public lab, not a classification theorem for all cubics.
- The conclusions are tied to the sampled square and the current Newton cutoff.
- “Hotter near-critical lane” means hotter under this finite-budget measurement, not a universal statement about every possible statistic.

Those caveats matter.
But they do not erase the main result.
The two asymmetric cubics are already different enough to block the lazy story.


## Adversarial check

A weak follow-up would have been a second asymmetric cubic that merely changed the colors while repeating the same dominance pattern.

This one earns its place because it changes **both** of the main readings:

1. basin ownership stays much more balanced,
2. the near-critical late band stays materially hotter.

If a future third cubic does not change one of those structural readings, it probably does not deserve a new sidecar.


## Best next move

The strongest next step is **not** to keep stacking random cubics.

A better follow-up would be one of these:

- one third cubic only if it reveals a genuinely new critical-set geometry,
- or one higher-budget persistence check on this split-critical case if the heat map still changes shape instead of merely cooling.

That keeps the repo in lab mode instead of drifting into a polynomial scrapbook.
