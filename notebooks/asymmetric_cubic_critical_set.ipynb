# Breaking the cubic symmetry

This notebook is the slower companion to `reports/asymmetric-cubic.md`.

The narrow question is the useful one: once the clean `z^3 - 1` symmetry is broken, which parts of the old reading survive and which ones were really artifacts of the unity family?


## The two cubics

Unity cubic:

```text
p_u(z) = z^3 - 1
```

Asymmetric cubic:

```text
p_a(z) = (z - 1)(z - (-0.9 + 1.05i))(z - (-0.15 - 0.25i))
```

The first has one repeated critical point at the origin. The second has two distinct off-center critical points. That is the structural break this pass is trying to make visible.


````python
import csv
from pathlib import Path

rows = list(csv.DictReader((Path('..') / 'art' / 'asymmetric-cubic-critical-set-comparison.csv').open()))
unity = [row for row in rows if row['polynomial'] == 'unity-cubic']
asym = [row for row in rows if row['polynomial'] == 'asymmetric-cubic']
[(row['band_index'], round(float(row['late_fraction']), 3), round(float(row['dominant_share']), 3)) for row in unity]
````

The unity cubic gives the old classical picture back immediately: the first critical-distance band is the hot one, and the basin split stays close to one third each for most of the square.


````python
[(row['band_index'], round(float(row['late_fraction']), 3), round(float(row['dominant_share']), 3)) for row in asym]
````

The asymmetric cubic changes two things at once:

- the late tail is no longer concentrated in one circular center halo
- one root starts owning much more of the sampled square than the others

So symmetry breaking does not just move the critical points. It changes the basin politics too.


````python
from newton_fractal_lab.cubic import unity_cubic, asymmetric_cubic, cubic_critical_points

for polynomial in [unity_cubic(), asymmetric_cubic()]:
    print(polynomial.name)
    print('  roots:')
    for root in polynomial.roots:
        print('   ', complex(round(root.real, 4), round(root.imag, 4)))
    print('  critical points:')
    for point in cubic_critical_points(polynomial):
        print('   ', complex(round(point.real, 4), round(point.imag, 4)))
````

## Takeaway

The repo started with a clean family for a reason: `z^n - 1` makes the global geometry easy to read. But that reading carries hidden symmetry assumptions.

This sidecar makes one of them explicit:

1. in the unity cubic, the repeated center critical point organizes the hottest late tail
2. in the asymmetric cubic, the critical set splits and the basin shares stop staying democratic
3. so the right public follow-up metric is distance to the nearest critical point, not just distance to the origin

That is enough to open a real asymmetric lane without turning the repo into a generic polynomial catalogue.
