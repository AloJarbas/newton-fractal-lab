# Asymmetric cubic opposition

This notebook is the slower companion to `reports/asymmetric-cubic-opposition.md`.
It reads the generated summary CSV and checks the centroid-flip claim directly.


```python
import csv
from pathlib import Path

rows = []
with (Path('..') / 'art' / 'asymmetric-cubic-opposition.csv').open() as handle:
    for row in csv.DictReader(handle):
        parsed = {key: (float(value) if key != 'polynomial' else value) for key, value in row.items()}
        rows.append(parsed)
rows
```

## Check which cubic actually flips

A real counterweight lane should keep the root centroid positive while pushing the late-tail centroid negative.


```python
[(row['polynomial'], row['root_centroid_x'], row['late_tail_centroid_x']) for row in rows]
```

## Compare where the late mass lives

The new claim is not just about one x-coordinate. It is also about where the late mass concentrates across the square.


```python
sorted([(row['polynomial'], row['left_late_share'], row['center_late_share']) for row in rows], key=lambda item: item[1], reverse=True)
```
