# Three cubic persistence atlas

This notebook is the slower companion to `reports/cubic-persistence-atlas.md`.
It reads the generated CSV and checks the middle-lane claim directly.


```python
import csv
from pathlib import Path

rows = []
with (Path('..') / 'art' / 'cubic-persistence-atlas.csv').open() as handle:
    for row in csv.DictReader(handle):
        parsed = {key: (float(value) if key not in {'polynomial'} else value) for key, value in row.items()}
        rows.append(parsed)
rows
```

## Sort by surviving center heat

The cleanest read is just to rank the three cubics by high-budget center-four late share.


```python
sorted([(row['polynomial'], row['high_center_late']) for row in rows], key=lambda item: item[1], reverse=True)
```

## Check the middle-lane claim

The split-critical cubic should sit between the unity cubic and the older asymmetric cubic on both surviving center heat and inner-band retention.


```python
by_name = {row['polynomial']: row for row in rows}
unity = by_name['unity-cubic']
asym = by_name['asymmetric-cubic']
split = by_name['split-critical-asymmetric-cubic']
print('high center', unity['high_center_late'], split['high_center_late'], asym['high_center_late'])
print('inner retention', unity['inner_band_retained_fraction'], split['inner_band_retained_fraction'], asym['inner_band_retained_fraction'])
```
